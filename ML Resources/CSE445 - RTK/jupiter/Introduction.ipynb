```python
def IsOdd(x): # defining a function; IsOdd: function name; 
                # x: Parameters are variables that are passed into the function
    if x%2 == 0:
        return False
    else: 
        return True
```

```python
print (IsOdd(8)) # invoking the defined function
print (IsOdd(9))
IsOdd(5)
```
